```python
m = int(input('m = '))
print(m)
n = int(input('n = '))
print(n)
fm = 1
for num in range(1, m+1):
    fm*=num
fn = 1
for num in range(1, n+1):
    fn*=num
fm_n = 1
for num in range(1, m-n+1):
    fm_n *= num
print(fm//fn//fm_n)
```

```python
def fac(num):
    result = 1
    for n in range(1, num+1):
        result *= n
    return result

m = int(input('m = '))
print(m)
n = int(input('n = '))
print(n)
print(fac(m) // fac(n) // fac(m-n))
```

```python
from random import randint
def roll_dice(n = 2):
    total = 0;
    for _ in range(n):
        total += randint(1, 6)
    return total

def add(a=0, b=0, c=0):
    return a+b+c

print(roll_dice())
print(roll_dice(3))
print(add())

print(add(1))
print(add(1, 2))
print(add(1, 2, 3))
print(add(c=50,a=100,b=200))
```

```python
def add(*args):
    total = 0
    for val in args:
        total += val
    return total
print(add())
print(add(1, 2))
print(add(1, 2, 3))
print(add(1, 2, 3, 4, 5))
```

```python
def isPrime(x):
    check = True
    for i in range(2, x):
        if x % i == 0:
            check = False
    if check:
        return 'Is prime'
    else:
        return 'Not prime'
        
if __name__ == '__main__':
    print('call isPrime()')
    x = int(input('x = '))
    print('%d %s' % (x, isPrime(x)))
```

```python
def gcd(x, y):
    (x, y) = (y, x) if x > y else (x, y)
    for factor in range(x, 0, -1):
        if x % factor == 0 and y % factor == 0:
            return factor
def lcm(x, y):
    return x*y//gcd(x, y)
```
